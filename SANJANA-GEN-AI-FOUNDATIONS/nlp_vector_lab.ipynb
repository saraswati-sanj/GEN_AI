```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df = pd.read_csv("Reviews.csv", nrows=50000)

print("Dataset Shape:", df.shape)
```

```python
light_reviews = df[
    df["Text"].str.contains(r"\blight\b", case=False, na=False)
]

print("Reviews containing 'light':", len(light_reviews))
```

```python
for idx, text in enumerate(light_reviews["Text"].head(50)):
    print(f"\nReview {idx}")
    print(text[:400])
    print("-" * 80)
```

```python
sentence_1 = light_reviews.iloc[5]["Text"]
sentence_2 = light_reviews.iloc[18]["Text"]
sentence_3 = light_reviews.iloc[27]["Text"]

sentences = [sentence_1, sentence_2, sentence_3]
```

```python
for i, sentence in enumerate(sentences, start=1):
    print(f"\nSentence {i}:")
    print(sentence)
```

```python
vectorizer = CountVectorizer()

dtm = vectorizer.fit_transform(sentences)

dense_matrix = dtm.toarray()

print("\nVocabulary:")
print(vectorizer.get_feature_names_out())

print("\nDense Matrix:")
print(dense_matrix)
```

```python
vec_1 = dense_matrix[0]
vec_2 = dense_matrix[1]
vec_3 = dense_matrix[2]

print("\nvec_1:")
print(vec_1)

print("\nvec_2:")
print(vec_2)

print("\nvec_3:")
print(vec_3)
```

```python

similarity_1_2 = np.dot(vec_1, vec_2)

similarity_1_3 = np.dot(vec_1, vec_3)
```

```python
print("DOT PRODUCT SIMILARITY SCORES")

print("Sentence 1 vs Sentence 2:", similarity_1_2)
print("Sentence 1 vs Sentence 3:", similarity_1_3)
```
